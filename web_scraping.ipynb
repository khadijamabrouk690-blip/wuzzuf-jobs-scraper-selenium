```python
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.common.keys import Keys
from selenium.webdriver.chrome.options import Options
import time
import pandas as pd
import random


options = Options()
options.add_argument("--disable-blink-features=AutomationControlled")
options.add_experimental_option("excludeSwitches", ["enable-automation"])
options.add_experimental_option('useAutomationExtension', False)
# options.add_argument("--headless=new")
options.add_argument("--window-size=1920,1080")
options.add_argument("--disable-gpu")
options.add_argument("--no-sandbox")
options.add_argument("--disable-dev-shm-usage")


driver = webdriver.Chrome(options=options)

driver.get('https://wuzzuf.net/jobs/egypt')

job_list = []
positions = ['Data Engineer' , 'Data Science' , 'AI Developer' , 'Machine Learning Engineer']

for position in positions:
    search_input = driver.find_element(By.NAME, "q")
    search_input.clear()
    search_input.send_keys(position)
    #time.sleep(2)

    driver.find_element(By.CSS_SELECTOR, "button[type='submit']").click()

    time.sleep(5)
    for i in range(3):
        time.sleep(3)
        jobs = driver.find_elements(By.CSS_SELECTOR, "div.css-pkv5jc")
        for job in jobs:
            job_title = job.find_element(By.XPATH, ".//h2/a").text
            job_company = job.find_element(By.CSS_SELECTOR, "a.css-ipsyv7").text
            job_location = job.find_element(By.CSS_SELECTOR, "span.css-16x61xq").text
            try:
                job_type = job.find_element(By.CSS_SELECTOR, "div.css-5jhz9n span").text
            except:
                job_type = "Not specified"
            try:
                job_experience = job.find_element(By.CSS_SELECTOR, "div > div.css-1rhj4yg > div:nth-child(2) > span").text
            except:
                job_experience = "Not found"
            job_url = job.find_element(By.CLASS_NAME, "css-o171kl").get_attribute("href")

            job_list.append({
                "Title": job_title,
                "Company": job_company,
                "Location": job_location,
                "Type": job_type,
                "Experience": job_experience,
                "URL": job_url
            })
        try:
            random_sleep = random.uniform(5,10)
            time.sleep(random_sleep)
            driver.find_element(By.XPATH, "//a[@aria-label='Next page']//i[@class='css-b80bsd efou2fk0']//*[name()='svg']").click()
        except:
            print("No more pages for position:", position)
            break

driver.quit()
df = pd.DataFrame(job_list)
df.shape[0]
```

Data Cleaning

```python
df.drop_duplicates(inplace=True)
```

```python
df["Company"] = df["Company"].str.strip().str.replace("-", "", regex=False).str.strip()
df["Experience"] = df["Experience"].str.replace("·", "", regex=False).str.replace(".", "", regex=False).str.strip()
```
